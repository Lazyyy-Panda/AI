```python
def create_board():
    return [' ' for _ in range(9)]

def display_board(board):
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
```

```python
def check_win(board, player):

    for i in range(0, 9, 3):
        if board[i] == board[i+1] == board[i+2] == player:
            return True

    for i in range(3):
        if board[i] == board[i+3] == board[i+6] == player:
            return True

    if board[0] == board[4] == board[8] == player:
        return True
    if board[2] == board[4] == board[6] == player:
        return True
    return False
```

```python
def play_game():
    board = create_board()
    current_player = 'X'
    players = {'X': 'Player 1', 'O': 'Player 2'}
    game_over = False
    moves_made = 0

    print("Welcome to Tic-Tac-Toe!")
    print("Player 1 is X, Player 2 is O.")
    print("Enter a position from 1-9 to place your symbol.")

    while not game_over:
        display_board(board)
        print(f"{players[current_player]}'s turn ({current_player})")

        while True:
            try:
                position = int(input("Enter position (1-9): ")) - 1
                if 0 <= position <= 8 and board[position] == ' ':
                    board[position] = current_player
                    moves_made += 1
                    break
                else:
                    print("Invalid position or position already taken. Please choose again.")
            except ValueError:
                print("Invalid input. Please enter a number between 1 and 9.")

        if check_win(board, current_player):
            display_board(board)
            print(f"Congratulations, {players[current_player]} ({current_player}) wins!")
            game_over = True
        elif moves_made == 9:
            display_board(board)
            print("It's a draw!")
            game_over = True
        else:
            current_player = 'O' if current_player == 'X' else 'X'

    print("Game Over.")
```

```python
play_game()
```

Vaccum Cleaner Agent


```python
while True:
    try:
        N = int(input("Enter the number of rooms (N): "))
        if N > 0:
            break
        else:
            print("Number of rooms must be positive.")
    except ValueError:
        print("Invalid input. Please enter an integer.")


rooms = []
print("\nEnter the initial status for each room (1 for dirty, 0 for clean):")
for i in range(N):
    while True:
        try:
            status = int(input(f"Status for Room {i+1} (1/0): "))
            if status in [0, 1]:
                rooms.append(status)
                break
            else:
                print("Invalid status. Please enter 1 (dirty) or 0 (clean).")
        except ValueError:
            print("Invalid input. Please enter an integer (1 or 0).")
```

```python
print("\nInitial status of rooms:")
for i, status in enumerate(rooms):
    print(f"Room {i+1}: {'Dirty' if status == 1 else 'Clean'}")

curr = 0

print("\n--- Starting Vacuum Cleaner Agent ---")
```

```python
while curr < N:
    print(f"\nAgent is in Room {curr+1}")
    if rooms[curr] == 1:
        print(f"Room {curr+1} is dirty. Cleaning...")
        rooms[curr] = 0
        print(f"Room {curr+1} is now clean.")
    else:
        print(f"Room {curr+1} is already clean.")
    curr += 1
    if curr < N:
        print("Moving to the next room...")

print("\n--- Vacuum Cleaner Agent Finished ---")
```

```python
print("\nFinal status of rooms:")
for i, status in enumerate(rooms):
    print(f"Room {i+1}: {'Dirty' if status == 1 else 'Clean'}")
```
